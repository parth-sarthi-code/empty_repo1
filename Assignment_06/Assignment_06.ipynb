```python
# (b) Perform basic preprocessing: lowercasing, tokenization, punctuation removal, and stopword filtering
stopwords_en = {
    'is', 'are', 'was', 'were', 'the', 'a', 'an', 'and', 'or', 'in', 'on', 'at', 
    'to', 'for', 'of', 'with', 'from', 'by', 'can', 'it', 'this', 'that'
}

def preprocess_english(text):
    text = text.lower()
    tokens = re.findall(r'[a-z]+', text)
    cleaned_tokens = [w for w in tokens if w not in stopwords_en]
    return cleaned_tokens

processed_q1 = [preprocess_english(s) for s in corpus_q1]

print("First 3 preprocessed sentences:")
for i, sent in enumerate(processed_q1[:3], 1):
    print(f"Sentence {i}: {sent}")
```

```python
# (c) & (d) Train Word2Vec using CBOW architecture
# Hyperparameters:
# vector_size = 50 (dense embedding dimensionality)
# window = 5 (context window size: 5 words left and right)
# min_count = 1 (include all words in corpus)
# sg = 0 (CBOW architecture; sg=1 would be Skip-gram)
# epochs = 100 (iterations over the corpus)

model_cbow = Word2Vec(
    sentences=processed_q1,
    vector_size=50,
    window=5,
    min_count=1,
    sg=0,
    epochs=100,
    seed=42
)

print("--- CBOW Model Hyperparameters ---")
print("Architecture: Continuous Bag of Words (CBOW, sg=0)")
print("Vector Dimensionality:", model_cbow.vector_size)
print("Context Window Size:", model_cbow.window)
print("Min Word Count:", model_cbow.min_count)
print("Training Epochs:", model_cbow.epochs)
```

```python
# (e) Display vocabulary size and vector dimensionality
vocab_words = list(model_cbow.wv.index_to_key)
print(f"Vocabulary Size: {len(vocab_words)}")
print(f"Vector Dimensionality: {model_cbow.vector_size}")
print(f"Sample vocabulary words: {vocab_words[:10]}")
```

```python
# (f) Display vector representation of at least 10 selected words
selected_words_q1 = [
    'learning', 'data', 'models', 'algorithms', 'neural', 
    'networks', 'training', 'prediction', 'computer', 'vision'
]

vectors_df = pd.DataFrame(
    [np.round(model_cbow.wv[w][:6], 4) for w in selected_words_q1],
    index=selected_words_q1,
    columns=[f"Dim_{i+1}" for i in range(6)]
)
print("First 6 dimensions of 10 selected word vectors (CBOW):")
print(vectors_df)
```

```python
# (g) Calculate Cosine Similarity for at least five word pairs
word_pairs_q1 = [
    ('neural', 'networks'),
    ('machine', 'learning'),
    ('data', 'analysis'),
    ('image', 'vision'),
    ('computer', 'prediction')
]

pair_sims_q1 = []
for w1, w2 in word_pairs_q1:
    sim = model_cbow.wv.similarity(w1, w2)
    pair_sims_q1.append({'Word 1': w1, 'Word 2': w2, 'Cosine Similarity': round(float(sim), 4)})

print("Word-Pair Cosine Similarities (CBOW):")
print(pd.DataFrame(pair_sims_q1))
```

```python
# (h) Identify top similar words for at least five selected words
top_words_to_check = ['learning', 'neural', 'data', 'prediction', 'vision']

for word in top_words_to_check:
    similar_words = model_cbow.wv.most_similar(word, topn=3)
    formatted = ", ".join([f"{w} ({sim:.3f})" for w, sim in similar_words])
    print(f"Top 3 similar to '{word}': {formatted}")
```

### (i) Observations on Q1: The Distributional Hypothesis and CBOW Architecture

1. **Distributional Hypothesis:** Formulated by J.R. Firth (1957) — *"You shall know a word by the company it keeps."* Words that consistently appear in similar surrounding contexts share similar semantic meanings.
2. **CBOW Mechanism:** The Continuous Bag of Words (CBOW) architecture takes the context words within a sliding window (e.g. 5 words before and 5 words after) and attempts to predict the target center word. The word vectors are learned as the weights of the hidden projection layer.
3. **Learned Representations:** Words that frequently appear together (like `neural` and `networks`, or `machine` and `learning`) attain high cosine similarity in the vector space, proving that dense distributed representations successfully capture contextual semantics.

## Q2. Study GloVe and Compare Word Representations
In this task, we load pre-trained GloVe (Global Vectors for Word Representation) embeddings, compare them with our custom-trained Word2Vec (CBOW) model, examine differences in training methodologies, and test vector arithmetic for word analogies.

```python
# (d) Calculate Cosine Similarity for at least five word pairs using GloVe vectors
glove_pair_sims = []
for w1, w2 in word_pairs_q1:
    sim = glove_model.similarity(w1, w2)
    glove_pair_sims.append({'Word 1': w1, 'Word 2': w2, 'GloVe Cosine Similarity': round(float(sim), 4)})

print("Word-Pair Cosine Similarities (GloVe):")
print(pd.DataFrame(glove_pair_sims))
```

```python
# (e) Compare nearest neighbours obtained using Word2Vec (CBOW) and GloVe
comparison_records = []
for word in top_words_to_check:
    w2v_sim = [w for w, _ in model_cbow.wv.most_similar(word, topn=3)]
    glove_sim = [w for w, _ in glove_model.most_similar(word, topn=3)]
    comparison_records.append({
        'Target Word': word,
        'Word2Vec (CBOW) Top 3': ", ".join(w2v_sim),
        'GloVe (Pre-trained) Top 3': ", ".join(glove_sim)
    })

print("Nearest Neighbour Comparison (Word2Vec vs GloVe):")
print(pd.DataFrame(comparison_records))
```

```python
# (h) Word Analogy Experiment using Vector Arithmetic: King - Man + Woman = Queen
analogy_pairs = [
    (['king', 'woman'], ['man']),        # king - man + woman = queen
    (['paris', 'italy'], ['france']),    # paris - france + italy = rome
    (['computer', 'software'], ['hardware']) # software analogy
]

print("Word Analogy Vector Arithmetic (positive_words - negative_words):")
for pos, neg in analogy_pairs:
    predicted = glove_model.most_similar(positive=pos, negative=neg, topn=1)[0]
    formula = f"{pos[0]} - {neg[0]} + {pos[1]}"
    print(f"{formula} = {predicted[0]} (similarity: {predicted[1]:.4f})")
```

### Observations on Q2: Word2Vec vs GloVe Differences

1. **Local Context vs Global Co-occurrence:**
   - **Word2Vec (CBOW):** Predictive model that scans text using a local sliding context window and updates weights via gradient descent (local information).
   - **GloVe:** Count-based model that first builds a global word-word co-occurrence matrix over the entire corpus and factorizes the log-probability matrix using weighted least squares (global statistics).
2. **Corpus Size Impact:**
   - Word2Vec trained on a small 50-sentence technical corpus reflects domain-specific associations from that dataset.
   - GloVe trained on 6 billion tokens from Wikipedia captures universal world knowledge and broad semantic relationships (e.g. capitals, gender analogies).
3. **Vector Arithmetic:** GloVe linearizes semantic relationships into vector offsets, allowing analogies like $\vec{v}(\text{king}) - \vec{v}(\text{man}) + \vec{v}(\text{woman}) \approx \vec{v}(\text{queen})$.

```python
# (b) Apply Unicode normalization (NFC), tokenization, and basic preprocessing
punct_chars = "।,;:!?()[]{}|./«»'\""
hindi_stopwords = {
    'है', 'हैं', 'का', 'की', 'के', 'में', 'पर', 'से', 'को', 'और', 
    'ने', 'हो', 'था', 'थी', 'थे', 'एक', 'ही', 'करते', 'रहा', 'रही', 'रहे'
}

def preprocess_hindi(text):
    # Unicode NFC normalization
    text = unicodedata.normalize('NFC', text)
    tokens = text.split()
    cleaned = [w.strip(punct_chars) for w in tokens if w.strip(punct_chars)]
    return [w for w in cleaned if w not in hindi_stopwords]

processed_hindi = [preprocess_hindi(s) for s in corpus_hindi]

print("First 3 preprocessed Hindi sentences:")
for i, sent in enumerate(processed_hindi[:3], 1):
    print(f"Sentence {i}: {sent}")
```

```python
# (c) Train Word2Vec embeddings for the Indic language (Hindi CBOW)
hindi_w2v = Word2Vec(
    sentences=processed_hindi,
    vector_size=50,
    window=4,
    min_count=1,
    sg=0,
    epochs=100,
    seed=42
)

print("--- Hindi Word2Vec (CBOW) Trained ---")
print("Vocabulary Size:", len(hindi_w2v.wv))
print("Vector Dimensionality:", hindi_w2v.vector_size)
```

```python
# (d) Select at least 10 frequent words and generate their dense vectors
selected_hindi_words = [
    'भारत', 'देश', 'विज्ञान', 'तकनीक', 'विद्यार्थी', 
    'पुस्तक', 'शिक्षा', 'खेल', 'क्रिकेट', 'मौसम'
]

hindi_vectors_df = pd.DataFrame(
    [np.round(hindi_w2v.wv[w][:6], 4) for w in selected_hindi_words],
    index=selected_hindi_words,
    columns=[f"Dim_{i+1}" for i in range(6)]
)
print("First 6 dimensions of 10 selected Hindi word vectors:")
print(hindi_vectors_df)
```

```python
# (e) Calculate similarity between at least five pairs of regional-language words
hindi_word_pairs = [
    ('भारत', 'देश'),
    ('विद्यार्थी', 'पुस्तक'),
    ('विज्ञान', 'तकनीक'),
    ('खेल', 'क्रिकेट'),
    ('मौसम', 'प्रकृति')
]

hindi_sims = []
for w1, w2 in hindi_word_pairs:
    sim = hindi_w2v.wv.similarity(w1, w2)
    hindi_sims.append({'Word 1': w1, 'Word 2': w2, 'Cosine Similarity': round(float(sim), 4)})

print("Hindi Word-Pair Cosine Similarities:")
print(pd.DataFrame(hindi_sims))
```

```python
# (f) Generate sentence representations using the average of word embeddings (Mean Pooling)
def get_sentence_vector(tokens, model):
    vectors = [model.wv[w] for w in tokens if w in model.wv]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vectors, axis=0)

hindi_sentence_vectors = np.array([get_sentence_vector(s, hindi_w2v) for s in processed_hindi])

print(f"Generated {len(hindi_sentence_vectors)} sentence vectors.")
print("Sentence Vector Shape:", hindi_sentence_vectors.shape)
print("Sample Sentence Vector (first 6 dims):", np.round(hindi_sentence_vectors[0][:6], 4))
```

### (h) & (i) Observations on Q3: Characteristics and Challenges in Regional Language Representations

1. **Word vs Sentence Representations:**
   - **Word Vector:** Represents a single lexical unit based on its immediate co-occurrence window (shape: `(50,)`).
   - **Sentence Vector:** Aggregates all word vectors in a sentence via mean pooling (shape: `(50,)`), capturing overall thematic content.
2. **Challenges in Regional/Indic Embeddings:**
   - **Morphological Complexity:** Indic languages are highly inflected (e.g. लड़का, लड़के, लड़कों). Without morphological stemmers, each variant gets its own vector, fragmenting data statistics.
   - **Spelling Inconsistencies:** Use of optional nuktas (`क़` vs `क`), zero-width joiners (ZWJ/ZWNJ), and varied matra orderings create distinct Unicode sequences for identical words.
   - **Low Resource Settings:** Training robust word embeddings requires hundreds of millions of tokens, which are scarce for many regional Indic languages.

## Q4. Parallel Corpus and Cross-Lingual Information Retrieval

### (i) Observations on Q4: Challenges in Cross-Lingual Retrieval and Aligned Embedding Spaces

1. **Unaligned Embedding Spaces Problem:**
   - If an English Word2Vec model and a Hindi Word2Vec model are trained independently on monolingual data, their vector spaces are completely disconnected and mutually orthogonal.
   - The cosine similarity between an English vector and a Hindi vector in unaligned spaces is zero or mathematically meaningless.
2. **How Aligned Spaces Bridge the Gap:**
   - **Bilingual Lexicon Mapping:** Maps semantic anchors from language A to language B, allowing sentences to share the same vector space.
   - **Cross-Lingual Projection (e.g. Procrustes / Linear Alignment):** Learns an orthogonal transformation matrix $W$ such that $X W \approx Y$ for translation pairs.
   - **Modern Multilingual Embeddings:** Modern transformer models (like mBERT, XLM-RoBERTa, LASER) train across 100+ languages simultaneously using shared subword vocabularies and parallel translation objectives, allowing cross-lingual search with zero manual lexicon mapping.
3. **Retrieval Performance:** With aligned representations, semantic similarity between translation pairs is maximized, resulting in 100% Recall@1 on our test queries.
